```python
paragraph=""" The quick brown foxes were jumping over the lazy dogs. 
They are often seen running across the park in the early morning. 
Some foxes are very quick and can leap from one side of the fence
to the other with ease. However, not all of them are fast; some of them
prefer to sit and watch the world go by. Foxes have been known to adapt
well to urban areas, finding food and shelter in various places."""
```

this is to see what are the parts o speeches inide  paragraph

```python
#Apply stopwords if you need 
from nltk.corpus import stopwords
```

```python
#first break the para into sentences
import nltk
sentences=nltk.sent_tokenize(paragraph)
```

```python
sentences
```

```python
#go through each sentence and break into words
stop_words = set(stopwords.words('english'))
for i in range (len(sentences)):
    words=nltk.word_tokenize(sentences[i])
    words=[word for word in words if word not in stop_words]
```

```python
 nltk.download('averaged_perceptron_tagger')
```

```python
pos_tag=nltk.pos_tag(words)
print(pos_tag)
```
