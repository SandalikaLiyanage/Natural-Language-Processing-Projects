```python
paragraph="""The quick brown foxes were jumping over the lazy dogs. 
They are often seen running across the park in the early morning. 
Some foxes are very quick and can leap from one side of the fence
to the other with ease. However, not all of them are fast; some of them
prefer to sit and watch the world go by. Foxes have been known to adapt
well to urban areas, finding food and shelter in various places.
"""
```

stopwords will remove the words like she,he, i, the ,of,to , why, their....

```python
from nltk.corpus import stopwords
```

```python
import nltk
nltk.download('punkt')
nltk.download('stopwords')
```

```python
stopwords.words('English')
```

```python
#initialize the stemmer
from nltk.stem import PorterStemmer 
stemmer=PorterStemmer()
```

```python
#perform the tokenization(breaking into senetences)
sentences =nltk.sent_tokenize(paragraph)
```

```python
#apply stopwords , filter and apply stemming
for i in range(len(sentences)):
     # Tokenize the sentence into words
    words = nltk.word_tokenize(sentences[i])
    
    # Filter stopwords and apply stemming
    words = [stemmer.stem(word) for word in words if word.lower() not in set(stopwords.words('English'))]
    
    # Rebuild the sentence from the filtered and stemmed words
    sentences[i] = ' '.join(words)
    
```

since above methods of stemming does not give much good form of words let's try with lemmatization

```python
from nltk.stem import WordNetLemmatizer
lemmatizer=WordNetLemmatizer
```

```python
#apply stopwords , filter and apply stemming
for i in range(len(sentences)):
     # Tokenize the sentence into words
    words = nltk.word_tokenize(sentences[i])
    
    # Filter stopwords and apply stemming
    words = [lemmatizer.lemmatize(word,pos='v') for word in words if word.lower() not in set(stopwords.words('english'))]
    
    # Rebuild the sentence from the filtered and stemmed words
    sentences[i] = ' '.join(words)
```
