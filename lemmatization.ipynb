```python
import nltk
nltk.download('wordnet')
```

```python
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

# Without specifying POS
print(lemmatizer.lemmatize("going"))  # Output: "going"

# Specifying POS as verb
print(lemmatizer.lemmatize("going", pos="v"))  # Output: "go"
```

Explanation:
Without POS: It assumes "going" is a noun (e.g., "She has a going interest"), so it stays unchanged.
With POS="v": It correctly identifies the base form (lemma) of "going" as "go."

```python
words = ["eating", "going", "flies", "better", "ran", "eats", "children", "saw", "wrote", "studying"]

for word in words:
    print(word + "------->" + lemmatizer.lemmatize(word, pos="v"))
```
